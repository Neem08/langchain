```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_community.utilities import GoogleSerperAPIWrapper
search = GoogleSerperAPIWrapper()
```

```python
from langchain.agents import create_agent
llm = ChatGoogleGenerativeAI(model="gemini-3.6-flash")
```

```python
search.run("What is the CJP protest")
```

```python
res = llm.invoke("Hii what is the weather today in BHAYANDER and Maharashtra state")
res.text
```

```python
agent = create_agent(model = llm, tools =[search.run], system_prompt="You are a weather informer and you use tools to give answer based on location")
```

```python
query = "hi what is the weather today at bhayander east"
```

```python
res = agent.invoke({"messages":[{"role":"user", "content":query}]})
print(res['messages'][-1].text)
```
