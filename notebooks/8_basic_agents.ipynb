```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_core.tools import tool
from langchain.agents import create_agent
```

```python
from langchain_groq import ChatGroq
llm = ChatGroq(model="openai/gpt-oss-20b")
```

```python
@tool
def add_number(a:int,b:int):
   """
   this tool is to be used for adding and return addition of 2 numbers
   
   Args:
   a: Number one
   b: Number two
   """
   return a + b


@tool
def whoistheowner():
    """
    returns the name of the creator of the ai agent
    """
    return "Neemisha"

```

```python
ans = add_number.invoke({"a":20,"b":10})
print(ans)
```

```python
agent = create_agent(
    model=llm,
    tools=[add_number, whoistheowner],
    system_prompt="you are a teacher who uses tools to give answer")
```

```python
res = agent.invoke({"messages":[{"role":"user", "content": "what is 2+3 and who is the owner"}]})
# for chunks in res['messages'] :
    # print(res)
    # print("\n")
    
print(res['messages'][-1].content)
```
