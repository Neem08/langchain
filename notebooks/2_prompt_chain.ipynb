Prompt Template - Dynamic Prompts

```python
llm = ChatGoogleGenerativeAI(model="gemini-3.6-flash")
out = StrOutputParser()
```

```python
def transform_toLC(result:str):
    return result.upper()
```

```python
chain_prompts = ChatPromptTemplate.from_messages([
    {
    "role":"system", "content" :"you are a transalator and translate input in {lang} ?"
    },
          { "role": "user","content":"{query}"},
           ])
```

```python
res = llm.invoke(chain_prompts.invoke({"lang":"Hindi","query":"I Love Langchain"}))
res.text
```

```python
chains = chain_prompts | llm | out | transform_toLC
```

```python
res = chains.invoke({"lang": "Hinglish", "query":"Hi I am Neemisha I love LangChain"})
print(res)
```
